# 00 · Bell state

Prepare $|\Phi^+\rangle = (|00\rangle + |11\rangle)/\sqrt{2}$ with H then CNOT, and sample it.

```julia
using Yao

bell = chain(2, put(1 => H), control(1, 2 => X))
ψ = zero_state(2) |> bell
display(statevec(ψ))

samples = measure(ψ; nshots=1000)
Dict(s => count(==(s), samples) for s in unique(samples))
```
